```python
import torch

# 检查是否有可用的 GPU
if torch.cuda.is_available():
    device = torch.device("cuda")  # 选择一个 GPU
    print("GPU 可用")
    print("可用的 GPU 数量:", torch.cuda.device_count())
    print("当前使用的 GPU:", torch.cuda.current_device())
    print("GPU 名称:", torch.cuda.get_device_name(device))
else:
    device = torch.device("cpu")  # 如果没有 GPU，使用 CPU
    print("GPU 不可用")

# 接下来的代码将在 device 上运行（可能是 GPU 或 CPU）
```

```python
import torch

# 创建一个形状为 (128, 16) 的张量
o_tensor = torch.randn(3,6, 2)
print(o_tensor)

# 使用 repeat 函数进行复制
# repeat 的参数表示每个维度的重复次数
# 在第二个维度上重复 8 次，从而将原始 (128, 16) 复制成 (128, 128)
new_tensor = o_tensor.repeat(1,1, 3)
print(new_tensor)  # 输出应为 (128, 128)
```
